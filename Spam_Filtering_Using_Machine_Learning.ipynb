#**Introduction**
- Spam detection is an important application of machine learning and natural language processing (NLP) used to classify messages as Spam or Not Spam (Ham).

- Spam messages are unwanted or malicious messages sent in bulk for advertising, fraud, or phishing purposes.

###**Dataset Description**

The dataset contains text messages labeled as:

1. Spam (1): Unwanted promotional or fraudulent messages

2. Ham (0): Legitimate non-spam messages

```python
# import libraries - for data processing and machine learning.
import numpy as np
import pandas as pd
```

```python
# load dataset - for analysis and latin-1 encoding to handle special characters
df = pd.read_csv("/content/drive/MyDrive/ML-CSV/spam.csv", encoding='latin-1')
```

```python
df.head()
```

###**Data Preprocessing**
Cleans and prepares raw text data for accurate model training and prediction.

```python
# dropping null values - Remove missing values and unnecessary columns.
df = df.drop(["Unnamed: 2", "Unnamed: 3", "Unnamed: 4"], axis = 1)
```

```python
df.isnull().sum()
```

```python
# columns rename
df = df.rename(columns = {"v1":"target", "v2":"text"})
```

```python
df.head()
```

```python
df.info()
```

```python
# label encoding - Convert spam/ham text labels into numeric values.
df = df.replace({"ham":"0", "spam": "1"})
```

```python
df["target"].value_counts()
```

```python
df["text"].value_counts().head()
```

```python
# Input and Output
X = df['text']
y = df['target']
```

###**Feature Extraction**

Vectorization Techniques:-
1. CountVectorizer - Counts words
2. TF-IDF(Term Frequency-Inverse Document Frequency) - Gives importance to important words

```python
# # Convert text into numerical data using CountVectorizer
# from sklearn.feature_extraction.text import CountVectorizer

# cv = CountVectorizer()
# X_vector = cv.fit_transform(X)
```

```python
# Apply TF-IDF - Convert text messages into numerical feature vectors.
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
X_vector = tfidf.fit_transform(X)
```

##**Model Selection**

Choose suitable for model based on problem.

Like: 0, 1 -> Logistic Regression

```python
# Train Test Split - Divide data into training and testing sets.
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_vector, y, test_size=0.2, random_state=42)
```

```python
# GaussianNB works with dense array
X_train_dense = X_train.toarray()
X_test_dense = X_test.toarray()
```

###**Logistic Regression**

- It is supervised machine learning technique algorithm used for classification (Yes / No, 0 / 1, Spam / Not Spam).

- It predicts the probabilities that input belong which class.

- It is commonly used in binary classification.

```python
from sklearn.linear_model import LogisticRegression

model_lr = LogisticRegression()
model_lr.fit(X_train_dense, y_train)
```

```python
# Prediction - Predict whether messages are spam or not spam.
y_pred_lr = model_lr.predict(X_test_dense)
```

```python
# Check model accuracy and classification performance
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print('Accuracy:', accuracy_score(y_test, y_pred_lr))
```

```python
conf_matrix = confusion_matrix(y_test, y_pred_lr)
print("\nConfusion Matrix:\n",conf_matrix)

print('\nClassification Report:\n')
print(classification_report(y_test, y_pred_lr))
```

###**KNN**
- It is a supervised machine learning algorithm used for both classification and regression task.

```python
from sklearn.neighbors import KNeighborsClassifier

model_knn = KNeighborsClassifier(n_neighbors = 5)
model_knn.fit(X_train_dense, y_train)
```

```python
# Prediction - Predict whether messages are spam or not spam.
y_pred_knn = model_knn.predict(X_test_dense)
```

```python
# Check model accuracy and classification performance
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print('Accuracy:', accuracy_score(y_test, y_pred_knn))
```

```python
conf_matrix = confusion_matrix(y_test, y_pred_knn)
print("\nConfusion Matrix:\n",conf_matrix)

print('\nClassification Report:\n')
print(classification_report(y_test, y_pred_knn))
```

###**Gaussian Naive Bayes**

A probabilistic classifier based on Bayes theorem.

Advantages:

1. Fast training
2. Works well for text classification
3. Handles large datasets efficiently

```python
# Model - Train Gaussian Naive Bayes model using training data.
from sklearn.naive_bayes import GaussianNB

model_nb = GaussianNB()
model_nb.fit(X_train_dense, y_train)
```

```python
# Prediction - Predict whether messages are spam or not spam.
y_pred = model_nb.predict(X_test_dense)
```

##**Model Evaluation**

Check performance metrics are::-

1. **Accuracy Score:** Measures how many predictions are correct out of all predictions.

Formula:

Accuracy = TP + TN + FP + FN / TP + TN

2. **Precision:** Measures how many predicted spam messages are actually spam.

Formula:

Precision = TP + FP / TP
	​

3. **Recall:** Measures how many actual spam messages are correctly detected.

Formula:

Recall = TP + FN / TP

4. **F1 Score:** Measures the balance between Precision and Recall.

Formula:

F1 Score = 2 × Precision + Recall / Precision × Recall


```python
# Check model accuracy and classification performance
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print('Accuracy:', accuracy_score(y_test, y_pred))

conf_matrix = confusion_matrix(y_test, y_pred)
print("\nConfusion Matrix:\n",conf_matrix)

print('\nClassification Report:\n')
print(classification_report(y_test, y_pred))
```

##**New Data - Prediction**

```python
# Predict new messages and conclude whether they are Spam or Not Spam
new_messages = [
    "Congratulations! You won ₹50,000. Click now to claim reward.",
    "Project meeting tomorrow at 10 AM. Please bring report."
]

predictions = model_nb.predict(tfidf.transform(new_messages).toarray())

for i in range(len(new_messages)):
    if predictions[i] == "1":
        print(new_messages[i], "--> Spam")
    else:
        print(new_messages[i], "--> Not Spam")
```
